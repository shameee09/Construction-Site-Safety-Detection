```python
from ultralytics import YOLO

model = YOLO("yolo11n.pt")

print("YOLO model loaded successfully")
```

```python
results = model.train(
    data=r"D:\Construction safety\dataset\data.yaml",
    epochs=30,
    imgsz=512,
    batch=2,
    workers=0,
    project=r"D:\Construction safety\runs",
    name="helmet_vest_detection",
    pretrained=True
)
```

```python
import os
best_model_path = r"D:\Construction safety\runs\helmet_vest_detection\weights\best.pt"

print("Best model exists:", os.path.exists(best_model_path))
print(best_model_path)
```

```python
from ultralytics import YOLO

model = YOLO(r"D:\Construction safety\runs\detect\train-2\weights\best.pt")

results = model.predict(
    source=r"D:\Construction safety\dataset\test\images",
    conf=0.4,
    save=True
)

print("Prediction completed!")
```
